# Leyendo archivos CSV y TXT

En lugar de crear estruturas de Series o DataFrames desde cero o incluso desde secuencias o ndarrays del nucleo de Python, el uso tipico de Pandas se basa en la cargar de información desde archivos o fuentes de información para su posterior exploración, transformacion y análisis

En esta lectura aprenderemos como leer valores de archivos (.csv) separados por comas y archivos de filas de texto (.txt) con los DataFrames de Pandas

```python
import pandas as pd
```

Cuando quieres trabajar con un archivo los primero que tienes que hacer es abrirlo. Esto se hace invocando a la función **open()**, el cual tiene solo argumento requerido que es la ruta donde se encuentra el archivo y tiene un solo valor de retorno que es el objeto archivo

La instrucción **with** se encarga automáticamente de cerrar el archivo una vez que sale del bloque with, incluso en casos de error.

```python
with open('btc-market-price.csv') as fp:
    print(fp)
```

**Explicación del código de abajo**

Se abre el archivo y lo guarda en la variable **fp**. La palabra **with** es importante por que:
- Abre el archivo
- Lo usa dentro del bloque
- Lo cierra automáticamente al terminar

~~~python
with open('btc-market-price.csv') as fp:
~~~

- **fp.readlines** Esto lee todas las lineas del archivo y devuelve un lista de strings, donde cada elemento es una linea

- **enumarate(fp.readlines)** Esto sirve para obtener indice y valor al mismo tiempo

~~~python
    for index, line in enumerate(fp.readlines()):
        # read just the first 10 lines
        if(index < 10):
            print(index, line)
~~~

```python
with open('btc-market-price.csv') as fp:
    for index, line in enumerate(fp.readlines()):
        # read just the first 10 lines
        if(index < 10):
            print(index, line)
```

```python
with open('btc-market-price.csv') as fp:
    for index, line in enumerate(fp.readlines()):
        # read just the first 10 lines
        if(index < 10):
            timestamp, price = line.split(',')
            print(f"{timestamp}: ${price}")
```

```python
import csv
```

```python
with open('exam_review.csv', 'r') as fp:
    reader = csv.reader(fp, delimiter = '>') # delimitador especial
    next(reader) # Saltando la cabecera
    for index, values in enumerate(reader):
        if not values:
            continue # Salta lineas vacias
        fname, lname, age, math, french = values
        print(f"{fname} {lname} (age {age}) got {math} in Math and {french} in French")
```
