# Mean-Variance-Standard Deviation Calculator

Crea una función llamada calculate() en **mean_var_std.py** que utilice Numpy para generar la media, la varianza, la desviación estándar, el máximo, el mínimo y la suma de las filas, columnas y elementos de una matriz de 3 x 3.

La entrada de la función debe ser una lista que contenga 9 dígitos. La función debe convertir la lista en una matriz Numpy de 3 x 3 y luego devolver un diccionario que contenga la media, la varianza, la desviación estándar, el máximo, el mínimo y la suma a lo largo de ambos ejes y para la matriz aplanada.

```python
import numpy as np
```

```python
def calculate(a):
    matrix3x3 = np.array([a[0:3], a[3:6], a[6:9]])
    return(
        {
            'mean': [matrix3x3.mean(axis = 0), matrix3x3.mean(axis = 1), matrix3x3.mean()],
            'variance': [matrix3x3.var(axis = 0), matrix3x3.var(axis = 1), matrix3x3.var()],
            'max': [matrix3x3.max(axis = 0), matrix3x3.max(axis = 1), matrix3x3.max()],
            'min': [matrix3x3.min(axis = 0), matrix3x3.min(axis = 1), matrix3x3.min()],
            'sum': [matrix3x3.sum(axis = 0), matrix3x3.sum(axis = 1), matrix3x3.sum()]
        }
    )

calculate([0,1,2,3,4,5,6,7,8])
```
