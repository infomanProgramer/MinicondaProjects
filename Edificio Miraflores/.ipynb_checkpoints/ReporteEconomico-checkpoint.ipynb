```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_julio_planilla = pd.read_csv(
    "planillas/julio_domestico.csv",
    sep=";",
    decimal=",",
    encoding="latin-1"
)
```

```python
df_julio_planilla.info()
```

```python
df_julio_planilla['PAGO?'] = df_julio_planilla['PAGO?'].fillna('')
df_julio_planilla
```

```python
nombres_excluir = ["CIDAL", "LAVANDERIA", "GIMNASIO"]

df_julio_planilla = df_julio_planilla[
    ~df_julio_planilla["NOMBRE Y APELLIDO"].str.contains(
        "|".join(nombres_excluir),
        case=False,
        na=False
    )
]
#Lo que se deberia pagar sin CIDAL ni espacios comunes como lavanderia ni gimnacio
total_area_domestica_julio = round(df_julio_planilla["TOTAL"].sum(), 2)
#print(total_area_domestica_julio)
suma_pagado_julio = 0
cont = 1
for indice, fila in df_julio_planilla.iterrows():
    cont += 1
    if fila["PAGO?"].strip() == "S" or fila["PAGO?"].strip() == "s":
        suma_pagado_julio += fila["TOTAL"]
    else:
        pago_str = str(fila["PAGO?"]).strip()
        if len(pago_str) > 0:
            pago_str = pago_str.replace(",", ".")
            suma_pagado_julio += float(pago_str)
            print(pago_str)
print(f"Total a pagar {total_area_domestica_julio}")
print(f"Total pagado {round(suma_pagado_julio, 2)}")
```
