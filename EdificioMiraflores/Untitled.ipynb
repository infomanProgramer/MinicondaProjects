```python
import numpy as np
import pandas as pd 
```

```python
def promedio(a, b, c):
    return round((a+b+c)/3)
```

```python
df_alumnos = pd.DataFrame({
    "Nombres": ["Rafael", "Juan", "Carlos"],
    "Apellidos": ["Ortiz Rocha", "Vargas", "Villegas"]
})

df_curso = pd.DataFrame({
    "Curso": ["Computacion", "Psicologia", "Matemáticas"],
    "Modalidad": ["Virtual", "Virtual", "Presencial"]
})

df_notas = pd.DataFrame({
    "1er Parcial": [58, 90, 45],
    "2er Parcial": [60, 85, 35],
    "3er Parcial": [58, 90, 80]
})

# df_notas["suma"] = df_notas["1er Parcial"] + df_notas["2er Parcial"] + df_notas["3er Parcial"]

df_notas["promedio"] = df_notas.apply(promedio(df_notas["1er Parcial"], df_notas["2er Parcial"], df_notas["3er Parcial"]))

tabla = pd.concat(
    [
        df_alumnos, 
        df_curso,
        df_notas
    ], axis = 1
).fillna(0)

tabla
```

```python
tabla.index.name = "Descripcion"
tabla
```

```python
tabla.loc["Total"] = tabla.sum()
tabla
```
