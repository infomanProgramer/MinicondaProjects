```python
df_planilla_v2['NRO.\nPARQUEO'] = (
    pd.to_numeric(df_planilla_v2['NRO.\nPARQUEO'], errors='coerce')
    .astype('Int64')
    .astype('string')
    .fillna('')
)

df_planilla_v2['NRO.\nBAULERA'] = (
    pd.to_numeric(df_planilla_v2['NRO.\nBAULERA'], errors='coerce')
    .astype('Int64')
    .astype('string')
    .fillna('')
)

for indice, fila in df_planilla_v2.iterrows():
    if fila["NRO.\nPARQUEO"].strip() != "":
        print(df_planilla_sistema[df_planilla_sistema["Unidad"] == "PARQUEO-"+fila["NRO.\nPARQUEO"]])

    if fila["NRO.\nBAULERA"].strip() != "":
        print(df_planilla_sistema[df_planilla_sistema["Unidad"] == "BAULERA-"+fila["NRO.\nPARQUEO"]])

```
