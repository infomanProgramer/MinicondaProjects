```python
import pandas as pd
```

```python
dates = pd.date_range("20210101", periods=6, freq="d")
stock_prices = [100.0, 200.0, 300.0, 250.0, 150.0, 175.0]
stock_price_series = pd.Series(stock_prices, index=dates)

# stock_price_series

"""
El parámetro W-SUN significa:
    W -> frecuencia semanal
    SUN -> la semana termina el domingo
"""

stock_price_series = stock_price_series.resample('W-SUN').mean().rolling(window=3).mean()
stock_price_series
```

```python
stock_price_series = pd.Series(stock_prices, index=dates)
stock_price_series = stock_price_series.resample('W-SUN').mean().rolling(window=3).std()
stock_price_series
```

```python
stock_price_series = pd.Series(stock_prices, index=dates)
stock_price_series = stock_price_series.resample('W-SUN').mean()
stock_price_series.index = stock_price_series.index.tz_localize("UTC").tz_convert("US/Eastern")
stock_price_series
```
