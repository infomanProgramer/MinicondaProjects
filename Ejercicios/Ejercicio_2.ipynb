```python
import pandas as pd
```

```python
data = {"Product A": 100, "Product B": 200, "Product C": 300, "Product D": 400}
```

```python
price_series = pd.Series(data)
```

```python
price_series.info()
```

```python
threshold = 150
```

```python
price_series = price_series[price_series > threshold]
```

```python
price_series
```

```python
 def transform_price(price: float) -> float:
        return price * 0.8
```

```python
price_series = price_series.apply(lambda row: transform_price(row)) 
```
