# Función Rolling

La funcion **rolling()** en Pandas se usa para aplicar calculos sobre una *ventana movil* de datos. Es muy común en analisis de series de tiempo, por ejemplo para calcular promedios moviles, sumas, acumulados en intervalos, desviaciones estándar , etc

```python
import pandas as pd
```

```python
data = pd.Series([10, 20, 30, 40, 50])
```

En el siguiente ejemplo se toma ventanas de 3 valores consecutivos y calcula el promedio de cada ventana



Toma ventana de 3 valores consecutivos y calcula el promedio de cada ventana

Porque:

|Ventana|Cálculo|Resultado|
|-------|-------|---------|
|10|Insuficiente|NaN|
|10,20|Insuficiente|NaN|
|10,20,30|(10+20+30)/3|20|
|20,30,40|(20+30+40)/3|30|
|30,40,50|(30+40+50)/3|40|

```python
data.rolling(3).mean()
```

**Parámetro principal**

Define el tamaño de la ventana

~~~python
df.rolling(window=3)
~~~

También se puede escribir:

~~~python
df.rolling(3)
~~~

**Operaciones comunes después de rolling**

- mean()
- sum()
- min()
- max()
- std()
- var()
