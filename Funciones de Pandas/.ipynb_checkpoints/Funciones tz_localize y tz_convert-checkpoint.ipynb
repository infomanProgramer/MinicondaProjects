# Funciones tz_localize() y tz_convert()

**tz_localize():** Asigna una zona horaria (sin cambiar la hora)
Se usa cuando tus fechas no tienen zona horaria (son naive). La función solo etiqueta los datos una zona horaria sin modificar la hora

```python
import pandas as pd

dates = pd.date_range("2024-01-01 10:00", periods=3)

print(dates)
```

El resultado sin zona horaria



Ahora agregamos una zona horaria

```python
dates.tz_localize("UTC")
```

Las horas siguen siendo las mismas, solo se agrego UTC



**tz_convert():** Covierte la hora a otra zona horaria. Se usa cuando la fecha ya tiene zona horaria. La función ajusta la hora al nuevo huso horario

```python
dates.tz_localize("UTC").tz_convert("America/New_York")
```

- Antes: 10:00 UTC
- Ahora: 05:00 New York
