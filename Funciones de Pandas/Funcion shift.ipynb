# Funcion Shift

Desplaza los valores de una Serie o DataFrame hacia arriba o hacia abajo un número determinado de posiciones. Esto es muy común en series temporales, por ejemplo para comparar un valor con el anterior o el siguiente.

~~~python
shift(n)
~~~

- n > 0 -> desplaza los datos hacia abajo
- n < 0 -> desplaza los datos hacia arriba

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40])

print(s.shift(1))
```

```python
print(s.shift(-1))
```

**Ejemplo muy común en análisis de datos**

Caluclar la diferencia con el valor anterior

~~~python
df["difference"] = df["price"] - df["price"].shift(1)
~~~



Esto se usa mucho en:

- Análisis financiero
- Análisis de ventas
- Series temporales
