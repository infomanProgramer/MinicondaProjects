```python
import pandas as pd
```

## Método sort_index()

Ordena alfabeticamente por indice

```python
data = {"C": 30, "A": 10, "B": 20}
s = pd.Series(data)

s
```

```python
s.sort_index()
```

## Método sort_values()

Ordenar por valores

```python
data = {"C": 30, "A": 10, "B": 20}
s = pd.Series(data)

s.sort_values() # Ordena de forma ascendente
```

```python
s.sort_values(ascending = False)
```
