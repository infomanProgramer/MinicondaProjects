```python
import pandas as pd
from openpyxl import Workbook

vecinos = [
    {
        "departamento": "101",
        "nombre": "Juan Pérez",
        "deuda": {
            "jun-26": 261.71,
            "jul-26": 243.33,
            "ago-26": 226.72
        }
    },
    {
        "departamento": "102",
        "nombre": "María López",
        "deuda": {
            "mar-26": 192.27,
            "sep-26": 200.42
        }
    },
    {
        "departamento": "103",
        "nombre": "Carlos Ruiz",
        "deuda": {
            "nov-25": 180.00,
            "dic-25": 180.00,
            "ene-26": 190.00,
            "feb-26": 190.00
        }
    }
]

wb = Workbook()
ws = wb.active
ws.title = "Mora"

fila = 1

for vecino in vecinos:

    # Título del vecino
    ws.cell(fila, 1, 
            f"Departamento {vecino['departamento']} - "
            f"{vecino['nombre']}")
    fila += 1

    # Construir tabla de meses dinámicamente
    datos = {
        "Concepto": ["Expensas"],
        **{
            mes: [monto]
            for mes, monto in vecino["deuda"].items()
        }
    }

    df = pd.DataFrame(datos)
    columnas_meses = list(vecino["deuda"].keys())
    df["Total"] = df[columnas_meses].sum(axis=1)

    # Escribir encabezado y datos
    for col, nombre in enumerate(df.columns, start=1):
        ws.cell(fila, col, nombre)

    fila += 1

    for col, valor in enumerate(df.iloc[0], start=1):
        ws.cell(fila, col, valor)

    # Dejar dos filas de separación
    fila += 3

wb.save("reporte_mora.xlsx")
```
