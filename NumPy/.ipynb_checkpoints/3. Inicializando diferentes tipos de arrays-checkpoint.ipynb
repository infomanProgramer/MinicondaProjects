# Inicializando diferentes Arrays

```python
import numpy as np
import sys
```

```python
np.zeros(5) #  Arrays de una dimensión
```

```python
np.zeros((2, 3)) # Arrays de dos dimensiones
```

```python
np.zeros((2, 3, 3)) # Arrays de tres dimensiones
```

```python
np.zeros((2, 3, 3, 2)) # Arrays de cuatro dimensiones
```

```python
np.ones((4, 2, 2), dtype='int32') # Matriz de 3 dimensiones con valores 1
```

```python
# Creando una matriz de cualquier otro número
np.full((2, 2), 99, dtype='float32')
```

```python
# Números decimales randomicos
np.random.rand(4, 2)
```

```python
# Números enteros randomicos del 0 al 20
np.random.randint(20, size=(3, 3))
```

```python
# Números enteros randomicos entre 4 y 10
np.random.randint(4, 10, size=(3, 3))
```

```python
# Matriz identidad
np.identity(5)
```

```python
# Repetir un array
arr = np.array([1, 2, 3, 4])
r1 = np.repeat(arr, 3)
print(r1)
```

```python
# Repetir un array
arr = np.array([[1, 2, 3, 4]])
r1 = np.repeat(arr, 3, axis=0)
print(r1)
```

```python
output = np.ones((5, 5))
print(output)
```

```python
z = np.zeros((3, 3))
z[1, 1] = 9
print(z)
```

```python
# array[filas, columnas]
output[1: -1, 1: -1] = z
```

```python
output
```

#### Ser cuidadoso cuando se copian arrays

```python
a = np.array([1, 2, 3])
b = a
b
```

```python
b[0] = 100
b
```

```python
a # Como se ve la anterior modificación no solo cambio el array B sino que tambien modifico el array A
```

```python
# Para solucionar este problema podemos usar la función copy()

a = np.array([1, 2, 3])
b = a.copy()
b[0] = 100
print(a)
print(b)
```
