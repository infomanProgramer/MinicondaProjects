# Mas Visualizaciones

Previamente, vimos como el método **plot** de Pandas trabaja y como usa la API básica de Matplotlib. Daremos mas detalles en esta lección

**Matplotlib tiene dos APIs importantes que son:**
- **API Global** La primera es la más común y utilizada y esta inspirada en MATLAB
    - ¿Por que se considera Global?
      Por que estamos importando funciones a nivel de modulo. Por ejemplo *plt.figure* esta usando funciones directamente del modulo mayor *import matplotlib.pyplot as plt*
    - **OOP Interface** Y la segunda que esta mas orientada a objetos

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

## API Global

La API pyplot predeterminada de Matplotlib tiene una interfaza global al estilo de MATLAB como ya hemos visto

```python
x = np.arange(-10, 11) # Crea un arreglo de -10 a 10 (el 11 no se incluye)
x
```

- Crea una figura 12 pulgadas de ancho por 6 pulgadas de alto
~~~python
plt.figure(figsize = (12, 6))
~~~
- Le pone un título al gráfico
~~~python
plt.title('My Nice Plot')
~~~
- Esta line *plt.plot(x, x ** 2)* gráfica la función :

plt.plot(x, x ** 2)
- Esta linea *plt.plot(x, -1 * (x ** 2))* gráfica la función:

Es la misma parabola pero invertida hacia abajo

```python
# Todas la lineas de arriba en ejecución
plt.figure(figsize = (12, 6))
plt.title('My Nice Plot')
plt.plot(x, x ** 2)
plt.plot(x, -1 * (x ** 2))
```

## Veamos una ejemplo mas complicado

- Crea una figura de 12x6 pulgadas
~~~python
plt.figure(figsize = (12, 6))
~~~
- Coloca un título
~~~python
plt.title('My Nice Plot II')
~~~
- Acá se esta usando **subplot** (múltiples gráficos en una misma figura), a continuación explicamos que signfica cada parámetro;
    - 1 fila
    - 2 Columnas
    - Estoy trabajando en el gráfico número 1
~~~python
plt.subplot(1, 2, 1)
~~~
- Primer gráfico de la izquierda *y = x ^ 2*
~~~python
plt.plot(x, x ** 2)
~~~
- A continuación se dibuja una linea vertical en x = 0
    - Los valores de X son [0, 0, 0] -> siempre 0
    - Los valores de Y van de -10 a 100
    - Esto crea una linea vertical sobre el eje Y
x|y
-|-
0|-10
0|0
0|100
~~~python
plt.plot([0, 0, 0], [-10, 0, 100])
~~~
- Agrega la leyenda con los nombres de las dos curvas
~~~python
plt.legend(['X ^ 2', 'Vertical Line'])
~~~
- Agrega etiquetas a los ejes
~~~python
plt.xlabel('X')
plt.ylabel('X Squared')
~~~
- Ahora viene el segundo gráfico
- Misma estructura
    - 1 fila
    - 2 columnas
    - ahora en el gráfico número 2
    - Es el gráfico de la derecha
~~~python
plt.subplot(1, 2, 2)
~~~
- Gráfica *y = -x ^ 2*
~~~python
plt.plot(x, -1 * (x ** 2))
~~~
- Acá se crea una linea horizontal en Y = -50
    - X va de -10 a 10
    - Y siempre es -50
x|y
-|-
-10|-50
0|-50
10|-50
~~~python
plt.plot([-10, 0, 10], [-50, -50, -50])
~~~
- Agrega leyenda y etiquetas
~~~python
plt.legend(['-X ^ 2', 'Horizontal Line'])
plt.xlabel('X')
plt.ylabel('X Squared')
~~~

```python
plt.figure(figsize = (12, 6))
plt.title('My Nice Plot II')
# 1 Fila, 2 Columnas, Gráfico 1
plt.subplot(1, 3, 1)
# Primer gráfico
plt.plot(x, x ** 2)
# Linea vertical en X = 0. Los valores de X son [0, 0, 0], 
# los valores de Y van de -10 a 100. Esto crea una linea vertical sobre el eje Y
plt.plot([0, 0, 0], [-10, 0, 100])
plt.legend(['X ^ 2', 'Vertical Line'])
plt.xlabel('X')
plt.ylabel('X Squared')

plt.subplot(1, 2, 2)
plt.plot(x, -1 * (x ** 2))
plt.plot([-10, 0, 10], [-50, -50, -50])
plt.legend(['-X ^ 2', 'Horizontal Line'])

plt.xlabel('X')
plt.ylabel('X Squared')
```

## OOP Interface

- A continuación creamos varias cosas importantes, la funcón **subplots** devuelve dos objetos
    - **fig** es la figura completa (el lienzo)
    - **axes** es el area donde se dibuja el gráfico
    - **figsize = (12, 6)** tamaño en pulgadas
    - Si no se especifica filas/columnas por defecto se crea 1 gráfico
~~~python
fig, axes = plt.subplots(figsize = (12, 6))
~~~
- **Primera curva:**
    - Esta gráficando *Y = X ^ 2*
    - *color = 'red'* Linea roja
    - *linewidth = 3* Grosor de linea
    - *marker = '0'* Dibuja circulos en cada punto
    - *markersize = 8* Tamaño de los circulos
    - *label = 'X ^ 2'* Texto para la leyendo
~~~python
axes.plot(
    x, (x ** 2), color = 'red', linewidth = 3,
    marker = 'o', markersize = 8, label = 'X ^ 2')
~~~
- **Segunda curva:**
    - Esta gráficando *Y = -X ^ 2*
    - Acá se esta usando el formato corto 'b--'
    - **b** -> blue (azul)
    - **--** Linea discontinua
~~~python
# color = 'blue', linestyle = '--'
axes.plot(x, -1 * (x ** 2), 'b--', label = '-X ^ 2')
~~~
- Coloca nombres a los ejes
~~~python
axes.set_xlabel('X')
axes.set_ylabel('X Squared')
~~~
- Coloca el Título
~~~python
axes.set_title('My Nice Plot')
~~~
- Leyenda. Muestra la leyenda usando **label =** definidos en cada **plot()**
~~~python
axes.legend()
~~~
En Jupyter Notebook, simplemente escribir **fig** para que muestre la figura
~~~python
plt.show() # En scripts normales
fig # Muestra la figura en Jupyter Notebook
~~~

```python
fig, axes = plt.subplots(figsize = (12, 6))
```

```python
axes.plot(
    x, (x ** 2), color = 'red', linewidth = 3,
    marker = 'o', markersize = 8, label = 'X ^ 2')

axes.plot(x, -1 * (x ** 2), 'b--', label = '-X ^ 2')

axes.set_xlabel('X')
axes.set_ylabel('X Squared')

axes.set_title('My Nice Plot')

axes.legend()

fig
```

```python
fig, axes = plt.subplots(figsize = (12, 6))

axes.plot(x, x + 0, linestyle='solid')
axes.plot(x, x + 1, linestyle='dashed')
axes.plot(x, x + 2, linestyle='dashdot')
axes.plot(x, x + 3, linestyle='dotted')

axes.set_title('My Nice Plot')
```

```python
fig, axes = plt.subplots(figsize = (12, 6))

axes.plot(x, x + 0, '-og', label = 'solid green')
axes.plot(x, x + 1, '--c', label = 'dashes cyan')
axes.plot(x, x + 2, '-.b', label = 'dashdot blue')
axes.plot(x, x + 3, ':g', label = 'dotted red')

axes.set_title('My Nice Plot II')
axes.legend()
```

## Función subplot2grid

Esta es otra forma de hacer gráficos usando *grid* en formato como;

```python
plt.figure(figsize = (14, 6))

ax1 = plt.subplot2grid((3, 3), (0, 0), colspan = 3)
ax2 = plt.subplot2grid((3, 3), (1, 0), colspan = 2)
ax3 = plt.subplot2grid((3, 3), (1, 2), colspan = 2)
ax4 = plt.subplot2grid((3, 3), (2, 0))
ax5 = plt.subplot2grid((3, 3), (2, 1))
```

### Diagrama de dispersión

```python
N = 50
x = np.random.rand(N)
y = np.random.rand(N)
colors = np.random.rand(N)
area = np.pi * (20 * np.random.rand(N))**2
```

```python
plt.figure(figsize = (14, 6))

plt.scatter(x, y, s = area, c = colors, alpha = 0.5, cmap = 'Spectral')
plt.colorbar()
plt.show()
```

### Histogramas

```python
values = np.random.randn(1000)
```

```python
plt.subplots(figsize = (12, 6))
plt.hist(values, bins = 100, alpha = 0.8,
                 histtype = 'bar', color = 'steelblue',
                 edgecolor = 'green')
plt.xlim(xmin = -5, xmax = 5)
plt.show()
```
