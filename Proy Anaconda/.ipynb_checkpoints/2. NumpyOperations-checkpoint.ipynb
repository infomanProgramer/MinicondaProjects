# Boolean Arrays

```python
a = np.arange(4)
```

```python
a
```

```python
a[0], a[-1] # Primera forma: Muestra el primer y último elemento
```

```python
a[[0, -1]] # Muesra el primer y últimos elemento segunda forma
```

**Diferencias principales**
~~~python
a[0] # Devuelve un solo valor
a[[0, -1]] # Devuelve varios valores
~~~

a[[0, -1]] Esto es super útil cuando se quiere seleccionar elementos especificos sin usar slicing

**Indexación booleana**

Estamos usando una mascara booleana para seleccionar elementos

**Importante**

La longitud de la mascara debe coincidir con la del array, sino *NumPy* lanza un error

```python
a[[True, False, False, True]] # Muestra el primer y últimos elementos tercera forma
```

```python
a * 10
```

**a >= 2** Compara cada elemento del array con 2, de forma vectorizada, como se evalua:

- 0 >= 2 -> False
- 1 >= 2 -> False
- 2 >= 2 -> True
- 3 >= 2 -> True

```python
a >= 2
```

```python
a[a >= 2] # Filtra el array a y se queda solo con los valores mayores iguales a 2
```

```python
a.mean() # Devuelve el promedio del vector
```

```python
a[~(a > a.mean())] # Devuelve los elementos del vector que no son mayores al promedio
```

```python
a[(a > a.mean())] # Devuelve los elementos del vector que son mayores al promedio
```

```python
a[(a == 0) | (a == 1)] # Devuelve los elementos que son iguales a cero o uno
```

```python
a[(a <= 2) & (a % 2 ==0)] # Devuevle los número pares menores o iguales a 2
```

**Arrays aleatorios**

- **np.random.randint(100)** Esta sentencia de abajo genera enteros aleatorios desde 0 a 100 (100 no incluido)
- **size=(3,3)** Define la forma del array: 3 filas por 3 columnas

**Nota** Los números cambian cada vez que se ejecuta la celda

```python
A = np.random.randint(100, size=(3, 3))
```

```python
A
```

```python
A[np.array([
    [True, False, True],
    [False, True, False],
    [True, False, True]
])]
```

```python
A > 30
```

```python
A[A > 30]
```
