# Pandas - Introducción

Es importante saber la diferencia entre Series y DataFrame

La libreria de Pandas es útil para trabajar con datos estructurados. 

**¿Qué es un dato estructurado?**

Son datos que se almacenan en tablas como archivos CSV, hojas de calculo de Excel, tablas de base de datos. Datos no estructurados consiste en texto de forma libre, imagenes, sonido o video.

Seleccionar una sola columna de datos de un DataFrame (conjunto de datos) lo cual devuevle una Serie. Trabajando con objetos de una dimensión se hace facil mostrar como diferentes métodos y operadores trabajan. Muchas series devuelven otra serie como resultado. Esto lleva a la posibilidad de llamar a otros métodos sucesivamente, lo que se conoce como encadenamiento de métodos.

**Documentación oficial de Panda**

- http://pandas.pydata.org/pandas-docs/stable/ (Documentación oficial de Panda)
- https://github.com/PacktPublishing/Pandas-Cookbook-Second-Edition (Repositorio de Pandas 1.x Cookbook)

```python
import pandas as pd
import numpy as np
```

```python
g7_pop = pd.Series([34.467, 63.951, 80.940, 60.665, 127.061, 64.511, 318.523])
```

```python
g7_pop
```

Una seria de Pandas no tiene etiquetas de columna, solo es una columna de un *DataFrame*. Una serie tiene etiquetas de fila

```python
g7_pop.name = "Población G7 en millones"
```

```python
g7_pop
```

```python
g7_pop.dtype # devuelve el tipo de datos almacenados
```

```python
g7_pop.values # devuelve los valores del array
```

```python
type(g7_pop.values) # devuelve el tipo de valor con python nativo
```

```python
g7_pop.index # devuelve el rando de indices
```

```python
# Asinamos nombres a los indices para que se vuelva una lista ordenada
g7_pop.index = [
    'Canada',
    'France',
    'Germany',
    'Italy',
    'Japan',
    'United Kingdom',
    'United States'
]
```

```python
g7_pop
```

# Indexación con Pandas y Selección condicional

```python
g7_pop['Canada']
```

```python
g7_pop['Japan']
```

## Las posiciones númericas tambien pueden ser utilizadas con el atributo *iloc*

```python
g7_pop.iloc[0]
```

## Seleccionando múltiples elementos en uno

```python
g7_pop[['Italy', 'France']]
```

```python
g7_pop.iloc[[0, 1]]
```

## Slice

```python
l = ['a', 'b', 'c']
```

```python
l
```

```python
l[:2] # para este ejemlo con un array nativo de Python vemos que no se muestra el elemento de indice 2 que seria 'c'
```

```python
g7_pop['Canada': 'Italy'] # en cambio la serie en Pandas devuelve el elemento del indice 'Italy' esa seria la pequeña diferencia con un array en Python
```

## Selección condicional

```python
g7_pop
```

```python
g7_pop > 70
```

```python
g7_pop[g7_pop > 70]
```

```python
g7_pop.mean()
```

```python
g7_pop[g7_pop > g7_pop.mean()]
```

## Operaciones y métodos

Serie soportan operaciones vectorizadas y funciones de agregación como Numpy

- ~ not
- | or
- & and

```python
g7_pop
```

```python
g7_pop * 1_000_000 # Cada valor lo multiplica por 1000000 sin necesidad de bucles
```

```python
g7_pop[(g7_pop > g7_pop.mean() - g7_pop.std() / 2) | (g7_pop > g7_pop.mean() + g7_pop.std() / 2)]
```

```python
g7_pop.mean()
```

```python
np.log(g7_pop)
```

```python
g7_pop['France': 'Italy'].mean()
```

## Modifcando Series

```python
g7_pop['Canada'] = 40.5
```

```python
g7_pop
```

```python
g7_pop.iloc[-1] = 500
```

```python
g7_pop
```
