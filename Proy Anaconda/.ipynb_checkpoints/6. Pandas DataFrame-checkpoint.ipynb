## Indexación, Selección y Partición

```python
df
```

- **loc[]:** Acceda a un grupo de filas y columnas por etiqueta(s) o una matriz booleana. Las entradas permitidas son:
    - Una sola etiqueta como 5 o 'a' (tenga en cuenta que 5 se interpreta como una etiqueta del índice, y nunca como una posición entera a lo largo del índice)
    - Una lista de etiquetas ['a', 'b', 'c']
    - Un trozo de objeto con etiquetas ['a' : 'f']
    - Un array de booleanos de la misma longitud [True, False, True]
    - Una serie booleana alineable. El índice de la clave se alineará antes del enmascaramiento.
    - Un índice alineable. El índice de la selección devuelta será la entrada.
    - Una función invocable con un argumento (la serie o marco de datos que llama) y que devuelve un resultado válido para la indexación (uno de los anteriores)
    - Recibe dos parámetros separados por una coma, el primero son las filas a devolver y la segunda la columna df.loc['France': 'Italy', 'Population']. El resultado es una Serie Pandas con los valores de la columna *Pupulation*
    - La expresión df.loc['France': 'Italy', ['Population', 'GDP']] al igual que en el ejemplo anterior recibe dos parámetros, pero esta vez el segundo es un array de columnas. En este caso el resultado es un *DataFrame* no una *Serie*
- **iloc[]:** Se basa principalmente en la posición de números enteros desde 0 a length-1 del eje. Las entradas permitidas son:
    - Un entero por ejemplo [5]
    - Un array de enteros por ejemplo [4, 3, 0]
    - Una rebanada de objeto con [4:7]
    - Un array booleano
    - Una tupla de índices de filas y columnas. Los elementos de tupla constan de una de las entradas anteriores (0, 1)
    - La expresión df.iloc[1:3, 3] tiene dos parámetros donde el primero son las filas a devolver por posición y la segunda son las columnas por posición. El resultado para este ejemplo es una Serie Pandas
    - La expresión df.iloc[1:3, [0, 3]] al igual que en el ejemplo anterior recibe dos parámetros sin embargo en este caso particular el segundo parámetro recibe un array de filas seleccionadas. El resultado es un *DataFrame*

```python
df.loc['Canada'] # loc - Selecciona filas cuyo indice tiene la etiqueta 'Canada'. 
# Trabaja con etiquetas no con posiciones númericas
```

```python
df.iloc[-1] # iloc - Solo trabaja con posiciones númericas empezando desde cero 
```

```python
df['Population'] # Devuelve la columna Pupulation del DF 
```

**Nota:** En todos los casos ya sea filas o columnas los resultados devueltos on *Series* con su respectivo atribuo *Name*

## Eliminar elementos

**drop:** Elimina filas o columnas especificando nombres de etiquetas y el eje correspondiente
~~~python
DataFrame.drop(labels=None, *, axis=0, index=None, columns=None, level=None, inplace=False, errors='raise')
~~~

Parámetros:
- **label:** etiqueta única o iterable de etiqueta
- **axis:** 0 o 'index' para indices, 1 o 'columns' para columnas. El valor por defecto es 0
- **index:** etiqueta única o iterable de etiquetas
- **columns:** etiqueta única o iterable de etiquetas
- **level:** Entero o nombre de etiqueta, opcional
- **inplace:** Booleano, por defecto False. Para múltiples indices, nivel desde el cual las etiquetas serán eliminadas
- **errors:** {'ignore', 'raise'}, Por defecto 'raise'. Si 'ignore' suprime el error y solo se eliminan estiquetas existentes

Devuelve DataFrame o Ninguno DataFrame con las etiquetas de índice o columna especificadas eliminadas o Ninguno si inplace=True

```python
df
```

```python
df.drop('Canada') # Eliminando la fila de indice Canda
```

```python
df
```

```python
df.drop(['Canada', 'Japan']) # Eliminando dos filas
```

```python
df.drop(columns=['Population', 'HDI']) # Eliminacion de columnas
```

### Operaciones con Series

Creamos una serie Pandas llamada crisis que tiene los valores -1_000_000 (*el guión bajo es solo un separador visual en Python, el valor real es -1000000*) y -0.3 e indices 'GDP' 'HDI'

```python
crisis = pd.Series([-1_000_000, -0.3], index = ['GDP', 'HDI']) 
```

```python
crisis
```

```python
df[['GDP', 'HDI']]
```

Sumar la Serie *crisis* a un DataFrame, usando alineación automática por etiquetas (labels). Pandas no suma por posición, sino por nombre de columna (indice de la Serie)

**¿Que hace la siguiente expresión?**

- A cada valor de la columna GDP le suma -1_000_000
- A cada valor de la columna HDI le suma -0.3
- La operación se aplica fila por fila

Esto equivale a:

~~~python
df['GDP'] = df['GDP'] + crisis['GDP']
df['HDI'] = df['HDI'] + crisis['HDI']
~~~

pero de forma vectorizada y automática

```python
df[['GDP', 'HDI']] + crisis
```

En este ejemplo a pesar de las modificaciones previas que se realizarón si mostramos el DataFrame **df** nos lo sigue mostrando con los mismos valores iniciales, esto se debe a que las operaciones son inmutables, lo que quiere decir que no cambiaran el DataFrame subyacente. Por lo que tenemos que crear un nuevo DataFrame que almacene los resultados de la operación dada

```python
df
```

## Renombrando Columnas

**rename:** ¿Qué hace?

- Cambiar nombres de columnas
- Cambiar valores de indices
- Corregir errores tipográficos
- Estandarizar nombres (por ejemplo, pasar a mayusculas)

~~~python
DataFrame.rename(
    mapper=None, 
    *, 
    index=None, 
    columns=None, 
    axis=None, 
    copy=<no_default>, 
    inplace=False, 
    level=None, 
    errors='ignore'
)
~~~

- **mapper:** Diccionario o función para renombrar etiquetas se usa junto con *axis*
~~~python
df.rename({'GDP': 'Gross Domestic Product'}, axis=1)
~~~
- **index:** Diccionario o función para renombrar el indice (filas)
~~~python
df.rename(index = {'France': 'FR'})
~~~
- **columns:** Diccionario o función para renombrar columnas
~~~python
df.rename(columns={'GDP': 'Gross Domestic Product'})
~~~
- **axis:** Indica si el mapper se aplica a:
    - 0 o 'index' -> filas
    - 1 o 'columns' -> columnas
~~~python
df.rename({'GDP': 'GNP'}, axis='columns')
~~~
- **inplace:** 
    - True -> Modifica el DataFrame original
    - False (default) -> devuelve uno nuevo
~~~python
df.rename(columns={'GDP': 'GNP'}, inplace=True)
~~~
- **errors:**
    - 'ignore' (default); si la etiqueta no existe, no pasa nada
    - 'raise' : lanza error (keyError)
~~~python
df.rename(columns={'ABC': 'XYZ'}, errors='raise')
~~~
- **level:**
    - Se usa cuando el indice o columnas son ***MultiIndex***
    - Indica que nivel se quiere renombrar
- **copy:** Controla se se hace copia de los datos (uso interno / avanzado)

```python
df.rename(
    columns={
            'HDI': 'Human Development Index',
            'Anual Popcorn Consumption': 'APC'
        }, index = {
            'United States': 'USA',
            'United Kingdom': 'UK',
            'Argentina': 'AR'
        }
)
```
