# Conversió a binario de un numero decimal

```python
def decimal_a_binario(numero):
    if numero == 0:
        return "0"
    
    binario = ""
    while numero > 0:
        residuo = numero % 2
        binario = str(residuo) + binario
        numero = numero // 2
    
    return binario

# Ejemplo
a = decimal_a_binario(1023)
print(a)
print(len(a))

b = decimal_a_binario(1024)
print(b)  # Output: 101010
print(len(b))
```
