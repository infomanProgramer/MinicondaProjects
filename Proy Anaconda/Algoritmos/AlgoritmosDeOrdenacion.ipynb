# Algortimos de Ordenación

## Ordenación por Inserción

```python
array = [34, 45,22, 55, 20, 60, 80, 80, 100]

for i in range(1, len(array)):
    key = array[i]
    j = i-1
    while j >= 0 and array[j] < key:
        array[j+1] = array[j]
        j = j-1
    array[j+1] = key
print("Algoritmo Ordenado ascendentemente")
print(array)
```
