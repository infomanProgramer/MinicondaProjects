```python
import pandas as pd
```

```python
data = {
    "Sales": [1000000, 20000],
    "Marketing": [500000],
    "Engineering": [2000000],
    "Human Resources": [600000, 30000],
    "Finance": [800000],
}

revenue_ranges = [
    (0, 500000),
    (500000, 1000000),
    (1000000, 2000000),
    (2000000, 3000000),
]
```

```python
revenue_series = pd.Series(data)
```

```python
revenue_series
```

```python
revenue_series = revenue_series.apply(sum)
```

```python
revenue_series
```

```python
def get_revenue_range(revenue: int):   
        for x in revenue_ranges:
            if revenue >= x[0] and revenue <= x[1]:
                return str(x[0]) + "-" + str(x[1])
```

```python
print(get_revenue_range(500))
```

```python
revenue_series = revenue_series.apply(lambda row: get_revenue_range(row))
```

```python
revenue_series
```

```python
revenue_series.groupby(revenue_series).count()
```
