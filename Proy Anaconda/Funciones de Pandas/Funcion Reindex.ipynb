# Reindex

Se usa en Pandas para cambiar (reorganizar, agregar o eliminar) los indices de un *DataFrame* o *Series*.
En otras palabras, te permite:
- Cambiar el orden de las filas
- Agregar nuevos indices
- Eliminar indices que no estén en la nueva lista
- Rellenar valores faltantes cuando agregas nuevos indices

**Sintaxis**

~~~python
df.reindex(labels=None, index=None, columns=None, fill_value=None, method=None)
~~~

- **index:** Nuevo orden o nuevos indices de filas
- **columns:** Nuevo orden o nuevas columnas
- **fill_value:** Valor para rellenar cuando no hay datos
- **method:** Método de relleno ('ffill', 'bfill')

```python
import pandas as pd
```

```python
data = {"A": 10, "B": 20, "C": 30}
s = pd.Series(data)

print(s)
```

### Reordenar indices

```python
s.reindex(["C", "A", "B"])
```

### Agregar nuevos indices

```python
s.reindex(['A', 'B', 'C', 'D'])
```

### Rellenar los valores nuevos

```python
s.reindex(['A', 'B', 'C', 'D'], fill_value = 0)
```

### Rellenar con método (muy usado en series temporales)

```python
s.reindex(['A', 'B', 'C', 'D'], method = 'ffill')
```
