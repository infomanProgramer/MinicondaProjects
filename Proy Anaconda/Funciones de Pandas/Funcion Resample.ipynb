# Función Resample

Se usa para cambiar de frecuencia de datos de series temporales (datos que tienen fechas u horas)

En otras palabras:

*Agrupa los datos por periodos de tiempo (dias, semana, mes, año, etc) permite aplicar operaciones como promedio, suma, maximo, etc*

```python
import pandas as pd
```

```python
dates = pd.date_range("20240101", periods=4, freq="D")
stock_prices = [100, 110, 120, 130]
stock_price_series = pd.Series(stock_prices, index=dates)
stock_price_series
```

```python
stock_price_series.resample('ME').mean()
```

**Requisito importante**

Para usar **resample()** el indice debe ser una **fecha** (DatetimeIndex)
