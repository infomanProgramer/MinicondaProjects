# Funcipon Appy()

Es un método en la libreria de Pandas para trabajar con *DataFrames* y *Series*. Sirve para aplicar una función personalizada a:
- Cada elemento de una Serie
- Cada fila o columna de un DataFrame

Es como decir *Toma esta función y ejecutala sobre cada valor (o fila o columna)

```python
import pandas as pd
```

### Aplicado a una Serie

```python
s = pd.Series([1, 2, 3, 4])

resultado = s.apply(lambda x: x*2)
print(resultado)
```

### Aplicado a un DataFrame

```python
df = pd.DataFrame({
    "A": [1, 2, 3],
    "B": [4, 5, 6]
})
```

Aplicar por columnas (axis = 0 columnas) por defecto

```python
df.apply(sum)
```

Aplicar por filas (axis = 1)

```python
df.apply(lambda row: row["A"] + row["B"], axis = 1)
```
