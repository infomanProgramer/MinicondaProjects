# Reorganizando arrays

```python
import numpy as np
import sys
```

```python
before = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])
print(before)

after = before.reshape((8, 1))
print(after)
```

```python
after = before.reshape((4, 2))
after
```

```python
after = before.reshape((2, 2, 2))
after
```

### Vectores de apilamiento vertical

```python
v1 = np.array([1, 2, 3, 4])
v2 = np.array([5, 6, 7, 8])

print(np.vstack([v1, v2]))

print(np.vstack([v1, v2, v2, v2]))
```

### Vectores de apilamiento horizontal

```python
h1 = np.ones((2, 4))
h2 = np.zeros((2, 2))
print(h1)
print(h2)
np.hstack((h1, h2))
```
