### Producto vectorial y producto matricial

**Ejemplo 1** Producto de un vector demanda y un vector de precio. Suponga que un fabricante producto cuatro articulos. La demanda de los articulos viene dada por el vector demanda d = (30, 20, 40, 10). El precio unitario que el fabricante recibe por los articulos esta dado por el 

vector de precios p = 20
                      15
                      18
                      40

Si se satisface la demanda, ¿que ingresos tendra el fabricante?

```python
d = np.array([30, 20, 40, 10]) # Matriz 1 x 4
p = np.array([[20, 15, 18, 40]]) # Matriz 4 x 1
print(d)
print(p)
print(d.shape)
print(p.shape)
```

```python
ganancia = d * p
```

```python
ganancia.sum()
```
