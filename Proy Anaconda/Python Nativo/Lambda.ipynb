Las funciones *lambda* en Python son funciones anonimas (sin nombre)  que se escriben en una sola linea, normalmente se usan como argumento de otra función

**Sintaxis**

~~~python
lambda argumentos: expresión
~~~
- Solo puede tener una expresión
- Devuelve automáticamente el resultado (no usa return)

**Ejemplo**

Función normal
~~~python
def sumar(a, b):
    return a + b
~~~
Lo mismo con lambda
~~~python
sumar = lambda a, b : a + b
print(sumar(3, 5))
~~~

**¿Cuando se usa mayormente?**

Principalmente se usa con funciones como:
- map()
- filter()
- sorted()
- reduce()
