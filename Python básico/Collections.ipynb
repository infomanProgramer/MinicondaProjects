# Collections

Python tiene multiples tipos de colecciones versatiles, cada uno con caracteristicas y capacidades diferentes. Las colecciones mas comunes son:
- Lists
- Tuples
- Dictionaries
- Sets

Pero hay una propiedad que es común a todas ellas y es que las colecciones en Python son *heterogeneas*, lo que quiere decir que soportan distintos tipos de datos.

## Lists

Las listas son secuencias ordenadas **mutables**. Podemos argumentar que son el tipo de colección mas comun 

```python
l = [3, "Hello World", True]
len(l)
```

```python
print(l[0])
print(l[1])
```

Los indices negativos son soportados tambien

```python
print(l[-1])
print(l[-2])
```

```python
l.append("Nuevo elemento")
l
```

## Tuples

Las tuplas son similares a las listas, pero con la gran diferencia que ells **son inmutables**. Lo que quiere decir que una vez que una tupla ha sido creada no se puede modificar mas adelante

```python
t = (3, "Hello world", True)
```

```python
t[0]
```

```python
t[-1]
```

## Dictionaries

Diccionarios son colecciones que almacenan clave valores. La clave (Key) deve ser un objeto inmutable, generalmente usamos strings como keys. Los diccionarios **son mutables** y lo mas importante no tiene orden

```python
user = {
    "name": "Mary Smith",
    "email": "mary@example.com",
    "age": 30,
    "suscribed": True
}

user
```

```python
user["email"]
```

## Sets

Conjuntos son colecciones no ordenadas, en el cual la única caracteristíca es que ellos solo tienen un elemento único

```python
s = {3, 1, 3, 7, 9, 1, 3, 1}
s
```

```python
s.add(10) # Añadiendo un elemento
s
```

```python
s.pop()
s
```

```python
s.pop()
s
```

```python
9 in s
```

## Iterando colecciones

```python
for elem in l:
    print(elem)
```

```python
for key in user:
    print(key.title(), '=>', user[key])
```

```python
for value in user.values():
    print(value)
```

```python
for key, value in user.items():
    print(key, value)
```

```python
for i in range(5):
    print(i)
```

## Excepciones

Las excepciones se generan en tiempo de ejecución cuando se produce una situación anormal en su programa. Las excepciones tambien pueden ser construidas y generadas por su código. Ejemplo de una excepción

```python
age = "30"

if age > 21:
    print("Allowed entrance")
```

```python
try:
    if age > 21:
        print("Allowed entrance")
except:
    print("Something went wrong")
```

La parte de la excepción tambien puede recibir parámetros con la excepción esperada

```python
try:
    if age > 21:
        print("Allowed entrance")
except TypeError:
    print("Something went wrong")
```
