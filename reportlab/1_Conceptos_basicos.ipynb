```python
from reportlab.pdfgen import canvas
from reportlab.lib.units import inch
```

```python
def hello(c):
    c.drawString(100,100,"Hello World")
c = canvas.Canvas("hello.pdf")
hello(c)
c.showPage() # Guarda la pagina actual del canvas
c.save() # Almacena el archivo y cierra el canvas
```

## Configuración del lienzo

Analizando los argumentos del constructor para el lienzo

~~~python
def __init__(self,filename,
             pagesize=(595.27,841.89),
             bottomup = 1,
             pageCompression=0,
             encoding=rl_config.defaultEncoding,
             verbosity=0,
             encrypt=None):
~~~

- El argumento **filename** controla el nombre del archivo PDF final. También puede pasar cualquier flujo binario abierto (como sys.stdout, la salida estándar del proceso de Python con una codificación binaria) y el documento PDF se escribirá en ese
- El argumento **Pagesize** es una tupla de dos números. El lienzo por defecto se establece en tamaño $A4$. Los tamaños de pagina mas comunes los puedes encontrar en la libreria ***reportlab.lib.pagesize***, asi puedes usar expresiones como:

~~~python
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import letter, A4
myCanvas = canvas.Canvas('myfile.pdf', pagesize=letter)
width, height = letter  #keep for later
~~~

- El argumento **bottomup** cambia el sistema de coordenadas, aunque este argumento esta deprecado
- El argumento **pageCompression** determina si se comprime el flujo de operaciones PDF de cada página. Por defecto, los flujos de páginas no se comprimen, ya que la compresión ralentiza el proceso de generación del archivo. Si el tamaño de salida es importante, configure pageCompression=1, pero tenga en cuenta que los documentos comprimidos serán más pequeños, pero su generación será más lenta. Las imágenes siempre se comprimen, y esta opción solo ahorrará espacio si cada página contiene una gran cantidad de texto y gráficos vectoriales.

```python
def hello2(c):
    # move the origin up and to the left
    c.translate(inch,inch)
    # define a large font
    c.setFont("Helvetica", 14)
    # choose some colors
    c.setStrokeColorRGB(0.2,0.5,0.3)
    c.setFillColorRGB(1,0,1)
    # draw some lines
    c.line(0,0,0,1.7*inch)
    c.line(0,0,1*inch,0)
    # draw a rectangle
    c.rect(0.2*inch,0.2*inch,1*inch,1.5*inch, fill=1)
    # make text go straight up
    c.rotate(90)
    # change color
    c.setFillColorRGB(0,0,0.77)
    # say hello (note after rotate the y coord needs to be negative!)
    c.drawString(0.3*inch, -inch, "Hello World")
c = canvas.Canvas("hello2.pdf")
hello2(c)
c.showPage()
c.save()
```
